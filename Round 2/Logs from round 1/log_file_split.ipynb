```python
def split_log_file(input_path):
    # Output file paths
    sandbox_log_path = "sandbox_logs.log"
    activities_log_path = "activities_log.csv"
    trade_history_log_path = "trade_history.log"

    # Line ranges (inclusive)
    sandbox_start, sandbox_end = 1, 50001
    activities_start, activities_end = 50005, 80006
    trade_history_start = 80011  # to end

    with open(input_path, "r") as infile, \
         open(sandbox_log_path, "w") as sandbox_file, \
         open(activities_log_path, "w") as activities_file, \
         open(trade_history_log_path, "w") as trade_history_file:

        for i, line in enumerate(infile, start=1):
            if sandbox_start <= i <= sandbox_end:
                sandbox_file.write(line)
            elif activities_start <= i <= activities_end:
                activities_file.write(line)
            elif i >= trade_history_start:
                trade_history_file.write(line)

    print(f"✅ Created:\n- {sandbox_log_path}\n- {activities_log_path}\n- {trade_history_log_path}")

# Replace with your actual log file path
split_log_file("raw_log_file.log")
```
