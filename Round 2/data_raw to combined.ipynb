File to convert all the data.csv files into a readable format and save them in a new folder

```python
import pandas as pd
import numpy as np
import math
import openpyxl
```

```python
# Load the raw data
prices_day_minus1 = pd.read_csv("data_old/prices_round_2_day_-1.csv", sep=';')
prices_day_0 = pd.read_csv("data_old/prices_round_2_day_0.csv", sep=';')
prices_day_1 = pd.read_csv("data_old/prices_round_2_day_1.csv", sep=';')
trades_day_minus1 = pd.read_csv("data_old/trades_round_2_day_-1.csv", sep=';')
trades_day_0 = pd.read_csv("data_old/trades_round_2_day_0.csv", sep=';')
trades_day_1 = pd.read_csv("data_old/trades_round_2_day_1.csv", sep=';')
```

```python
# trades data does not have a day column, so we add it manually per day
trades_day_minus1["day"] = -1
trades_day_0["day"] = 0
trades_day_1["day"] = 1

# Combine all the trades and prices data into one DataFrame
trades = pd.concat([trades_day_minus1, trades_day_0, trades_day_1], ignore_index=True)
prices = pd.concat([prices_day_minus1, prices_day_0, prices_day_1], ignore_index=True)
```

```python
# Add an overall time column to the trades and prices data so that we can distinguish between orders from different days
# The timestamp is in multiples of 100 starting from 0 to 1000000, so we multiply the day by 1000000 and add the timestamp to get a unique time for each order
trades['global_time'] = trades['day'] * 1000000 + trades['timestamp']
# Sort the trades by global time to ensure they are in the correct order
trades.sort_values(by='global_time', inplace=True)

# The prices data has a similar structure, so we do the same for it
prices['global_time'] = prices['day'] * 1000000 + prices['timestamp']
prices.sort_values(by='global_time', inplace=True)
```

```python
trades.to_excel("data/trades.xlsx", index=False)
prices.to_excel("data/prices.xlsx", index=False)
```
