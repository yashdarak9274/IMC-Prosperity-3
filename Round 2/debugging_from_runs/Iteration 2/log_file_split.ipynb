```python
import csv
import json

def split_and_convert_logs(input_path):
    # Output file paths
    activities_csv_path = "activities_log.csv"
    trade_history_csv_path = "trade_history.csv"

    # Line ranges
    activities_start, activities_end = 5006, 13006
    trade_history_start = 13012

    # Buffers
    activities_lines = []
    trade_lines = []

    # Step 1: Load selected lines into memory
    with open(input_path, "r") as infile:
        for i, line in enumerate(infile, start=1):
            if activities_start <= i <= activities_end:
                activities_lines.append(line)
            elif i >= trade_history_start:
                trade_lines.append(line)

    # Step 2: Convert and write activities log (semicolon-separated)
    with open(activities_csv_path, "w", newline="") as out_csv:
        reader = csv.reader(activities_lines, delimiter=';')
        writer = csv.writer(out_csv)
        for row in reader:
            writer.writerow(row)

    # Step 3: Convert and write trade history log (JSON list)
    if trade_lines and trade_lines[0].strip().startswith("Trade History"):
        trade_lines = trade_lines[1:]

    trade_data = json.loads("".join(trade_lines).strip())
    if trade_data:
        keys = trade_data[0].keys()
        with open(trade_history_csv_path, "w", newline="") as out_csv:
            writer = csv.DictWriter(out_csv, fieldnames=keys)
            writer.writeheader()
            writer.writerows(trade_data)

split_and_convert_logs("result.log")
```
